```python
# loading all the pakcages and the models
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv(('data/stud.csv'))
df.head()
```

```python
df['total_score'] = df['writing_score'] + df['reading_score'] + df['math_score']
df['total_score']

##let us first afhal add the average score as separate column
df['average'] = (df['math_score'] + df['reading_score'] + df['writing_score'])/3 
df.head()
```

```python
#now divide into dependent and independent features 
X = df.drop('total_score',axis=1)
y = df['total_score']
X
```

```python
#now divide into train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
X_train
```

```python
## now standardizing the dataset
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
oh_encoder = OneHotEncoder()

categorical_features = [feature for feature in X.columns if X[feature].dtype == 'O']
numerical_features = [feature for feature in X.columns if X[feature].dtype != 'O']

preprocessing = ColumnTransformer([
    ('onehotencoder',oh_encoder,categorical_features),
    ('numericalfeatures',scaler,numerical_features)
])

preprocessing
```

```python
X_train_scaled = preprocessing.fit_transform(X_train)
X_test_scaled = preprocessing.transform(X_test)
X_train_scaled
X_test_scaled
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
##this function returns the model score after training to check the performence of the model
def get_performence(y_test,y_pred):
    mse = mean_squared_error(y_test,y_pred)
    mae = mean_absolute_error(y_test,y_pred)
    r2_scores = r2_score(y_test,y_pred)
    return {"mse":mse,"mae":mae,"r2_score":r2_scores}
```

```python
models = [
LinearRegression(),
DecisionTreeRegressor(),
Ridge(),
Lasso(),
ElasticNet(),
SVR(),
RandomForestRegressor(),
GradientBoostingRegressor(),
AdaBoostRegressor(),
KNeighborsRegressor(),
XGBRegressor(),
CatBoostRegressor()    
]
```

```python
len(X_train_scaled)
len(y_train)
```

```python
for model in models:
    regressor = model.fit(X_train_scaled,y_train)
    y_train_predict = regressor.predict(X_train_scaled)
    scores_train_data = get_performence(y_test=y_train,y_pred = y_train_predict)

    print(f'------Score of {model} with respect to train data-------- \n')
    print(f'mse : {scores_train_data['mse']}')
    print(f'mae : {scores_train_data['mae']}')
    print(f'r2_score : {scores_train_data['r2_score']}')

    y_test_predict = regressor.predict(X_test_scaled)
    scores_test_data = get_performence(y_test=y_test,y_pred = y_test_predict)

    print(f'------Score of {model} with respect to test data-------- \n')
    print(f'mse : {scores_test_data['mse']}')
    print(f'mae : {scores_test_data['mae']}')
    print(f'r2_score : {scores_test_data['r2_score']}')

```

```python
###hyperparameter tuning
from sklearn.model_selection import GridSearchCV

linear_regresson = {
    'fit_intercept': [True, False],
    'positive': [True, False]}

decision_tree = {
    'max_depth': [3, 5, 10, 15, 20, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 4, 8],
    'max_features': [None, 'sqrt', 'log2']
}

ridge = {
    'alpha': [0.001, 0.01, 0.1, 1, 10, 100],
    'solver': ['auto', 'svd', 'cholesky', 'lsqr']
}

lasso = {
    'alpha': [0.0001, 0.001, 0.01, 0.1, 1, 10],
    'max_iter': [1000, 5000, 10000]
}

elastic_net = {
    'alpha': [0.0001, 0.001, 0.01, 0.1, 1],
    'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]
}

svr = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf', 'poly'],
    'gamma': ['scale', 'auto'],
    'epsilon': [0.01, 0.1, 0.2]
}

random_forest = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [1.0, 'sqrt', 'log2']
}

gradient_booster = {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [2, 3, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

adabooster = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.05, 0.1, 0.5, 1.0],
    'loss': ['linear', 'square', 'exponential']
}

kneighbors = {
    'n_neighbors': [3, 5, 7, 10, 15, 20],
    'weights': ['uniform', 'distance'],
    'p': [1, 2]
}

xgboosting = {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 5, 7],
    'min_child_weight': [1, 3, 5],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

cat_booster = {
    'iterations': [300, 500, 1000],
    'learning_rate': [0.01, 0.05, 0.1],
    'depth': [4, 6, 8, 10],
    'l2_leaf_reg': [1, 3, 5, 10]
}

models_with_params = [
    (LinearRegression(), linear_regresson),
(DecisionTreeRegressor(), decision_tree),
(Ridge(), ridge),
(Lasso(), lasso),
(ElasticNet(), elastic_net),
(SVR(), svr),
(RandomForestRegressor(), random_forest),
(GradientBoostingRegressor(), gradient_booster),
(AdaBoostRegressor(), adabooster),
(KNeighborsRegressor(), kneighbors),
(XGBRegressor(), xgboosting),
(CatBoostRegressor(), cat_booster)]


for label , param in models_with_params:
    grid = GridSearchCV(label,param_grid=param,scoring='neg_mean_squared_error',n_jobs=-1,cv=3)
    grid.fit(X_train_scaled,y_train)
    y_grid_pred = grid.predict(X_test_scaled)
    print(f'Best Params after {label} grid searh cv is {grid.best_params_}')
    print(f'Best Score after {label} grid searh cv is {grid.best_score_}')

    scores_test_data_grid = get_performence(y_test=y_test,y_pred = y_grid_pred)
    
    print(f'mse : {scores_test_data_grid['mse']}')
    print(f'mae : {scores_test_data_grid['mae']}')
    print(f'r2_score : {scores_test_data_grid['r2_score']}')

    print('\n')
```
